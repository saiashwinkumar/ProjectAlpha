```python
import os
import json
import boto3
from kafka import KafkaConsumer

# AWS session using your SSO profile
session = boto3.Session(
    profile_name="admin",
    region_name="ap-southeast-2"
)

s3 = session.client("s3")

# Kafka consumer
consumer = KafkaConsumer(
    'demo_testing2',
    bootstrap_servers=os.environ.get("KAFKA_BOOTSTRAP_SERVERS", "localhost:9092").split(","),
    value_deserializer=lambda x: json.loads(x.decode('utf-8'))
)

# Continuously consume Kafka messages and write them to S3
for count, message in enumerate(consumer):
    s3.put_object(
        Bucket='s3-stock-market-project-ashwin',
        Key=f'stock_market_{count}.json',
        Body=json.dumps(message.value),
        ContentType='application/json'
    )

    print(f"Uploaded stock_market_{count}.json")
```
